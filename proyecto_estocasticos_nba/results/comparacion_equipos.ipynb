## 2. Conclusiones

Todos los equipos fueron analizados con el mismo pipeline:
- Modelo Poisson (homogeneo y no homogeneo local/visitante)
- Pruebas de estacionariedad (ADF + KPSS)
- Distribucion exponencial de tiempos entre eventos de altos puntos
- Cadenas de Markov W/L con matriz de transicion y prediccion

Los resultados detallados de cada equipo estan en su carpeta individual
dentro de `results/`.
