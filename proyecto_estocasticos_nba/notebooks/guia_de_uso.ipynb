# Guía de Uso — Proyecto de Procesos Estocásticos NBA

## ¿Qué hace este programa?

Este programa analiza si la cantidad de puntos que anota un equipo de la NBA en cada partido se puede modelar con una **distribución de Poisson** (un concepto de probabilidad y estadística).

En palabras simples: ¿el número de puntos por partido se comporta como si fuera "aleatorio con una tasa promedio constante"?

### ¿Qué significan los resultados?

| Resultado | Significado |
|---|---|
| **Lambda** | Promedio de puntos por partido del equipo |
| **Razón Var/Media** | Si es ~1.0, hay buena señal de Poisson. Si es >1.2, hay sobredispersión |
| **P-valor > 0.05** | Los datos SÍ se ajustan a una distribución de Poisson |
| **P-valor < 0.05** | Los datos NO se ajustan a una distribución de Poisson |
